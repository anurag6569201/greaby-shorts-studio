```python
import os

AUDIO_FOLDER = '/Users/anuragsingh/Documents/GitHub/greaby-shorts-studio/src/assets/audio-models'

# Get all file names (filtering only audio files, e.g. .wav, .mp3, etc.)
audio_filenames = [
    f for f in os.listdir(AUDIO_FOLDER)
    if os.path.isfile(os.path.join(AUDIO_FOLDER, f)) and f.lower().endswith(('.wav', '.mp3', '.flac', '.aac', '.opus'))
]

# Print list of filenames
for filename in audio_filenames:
    print(filename)

# Optional: store in a Python list
print("\nAll filenames as list:\n")
print(audio_filenames)
```
